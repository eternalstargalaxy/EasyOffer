# DPO（Direct Preference Optimization）

> Rafailov et al., 2023. **绕过 RM**，把 RLHF 转成监督学习，直接用偏好对训练策略。

## 1. 题目背景
标准 RLHF（PPO）流程长且不稳：
1. 先训奖励模型 RM；
2. 再用 PPO 在线采样训练 actor + critic；
3. 4 个模型同时在线，显存大、超参敏感、训练易崩。

DPO 的关键洞察：**RLHF 的 KL 约束最优策略有闭式解**
$$\pi^*(y|x)=\frac{\pi_{ref}(y|x)\exp\big(r(x,y)/\beta\big)}{Z(x)}$$
把它反解出 $r(x,y)=\beta\log\frac{\pi^*(y|x)}{\pi_{ref}(y|x)}+\beta\log Z(x)$，
代入 Bradley-Terry 偏好模型 $P(y_w\succ y_l|x)=\sigma(r(x,y_w)-r(x,y_l))$，
**配分函数 $Z(x)$ 恰好消去**，得到只含 $\pi_\theta$ 和 $\pi_{ref}$ 的损失——**无需显式 RM**。

**面试常考点**：DPO 推导（闭式解 + BT 模型消 $Z$）、为什么无需 RM、
$\beta$ 调节、与 PPO 对比（offline/online、显存、稳定性）、reference model 作用。

## 2. 损失
给定偏好对 $(y_w\succ y_l\mid x)$（$y_w$ 胜 $y_l$）：
$$\mathcal L_{DPO}=-\log\sigma\!\left(\beta\log\frac{\pi_\theta(y_w|x)}{\pi_{ref}(y_w|x)}-\beta\log\frac{\pi_\theta(y_l|x)}{\pi_{ref}(y_l|x)}\right)$$
- $\pi_{ref}$：冻结的参考（SFT）策略，防偏离；
- $\beta$：温度，控制偏离 reference 的强度（$\beta$ 小→偏离大）；
- 直觉：让"胜者"的对数似然比相对 ref 升高，"败者"降低。

## 3. 与 PPO 对比
| 维度 | PPO | DPO |
|------|-----|-----|
| 训练范式 | online RL | offline 监督学习 |
| 需要 RM | 是 | 否（闭式解消去） |
| 需要 critic | 是 | 否 |
| 模型数量 | 4 (Actor/Critic/Ref/RM) | 2 (Policy/Reference) |
| 在线采样 | 需要 | 不需要 |
| 数据 | 在线 rollout | 静态偏好对 |
| 稳定性 | 不稳，超参敏感 | 稳定，监督学习 |
| 显存 | 高 | 中 |
| 探索能力 | 强（在线采样） | 弱（受限于偏好数据） |

## 4. 考察点
- 为什么无需 RM（RM 的最优解被代入策略闭式解，$Z(x)$ 消去）
- reference model 冻结与 $\beta$ 调节
- 与 PPO 对比：offline、无采样、无 critic，稳定但只能用已有偏好数据
- 隐式奖励 $r(x,y)=\beta\log\frac{\pi_\theta}{\pi_{ref}}$ 的含义


```python
# ===============================================================================
# DPO（Direct Preference Optimization）算法实现
# DPO通过人类偏好数据直接优化语言模型，使其生成更符合人类偏好的输出
# 这里面使用了一个偏好prefer以及两个reject的格式
# ===============================================================================
import torch
import torch.nn.functional as F
from transformers import LlamaForCausalLM, LlamaConfig
from copy import deepcopy

torch.manual_seed(0)
```

```python
# 加载模型
# 创建简化版的Llama模型作为策略模型（将被优化的模型）
policy_model = LlamaForCausalLM(config=LlamaConfig(vocab_size=12, num_hidden_layers=1, hidden_size=32))
# 创建参考模型（通常是SFT模型，在训练过程中保持不变）
reference_model = deepcopy(policy_model)  # 深度复制确保两个模型初始参数完全相同
policy_model
```

```python
# 超参数
beta = 0.1  # DPO的温度系数，控制策略模型与参考模型的偏离程度，值越小允许偏离越大

# 准备训练数据
# 在DPO中，我们需要提示(prompt)、优选回答(chosen/good)和拒绝回答(rejected/bad)
prompt_ids = [1, 2, 3, 4, 5, 6]  # 输入提示的token IDs
good_response_ids = [7, 8, 9, 2]  # 优质回答的token IDs
# 多个低质量回答的示例，每个都是token IDs的列表
bad_response_ids_list = [[1, 2, 0, 0], [4, 5, 6, 0]]
```

```python
# 构建模型输入：将提示与回答拼接
# 创建包含多个序列的批次：[提示+优质回答, 提示+低质回答1, 提示+低质回答2, ...]
input_ids = torch.LongTensor(
    [prompt_ids + good_response_ids, *[prompt_ids + bad_response_ids for bad_response_ids in bad_response_ids_list]]
)
input_ids
```

```python
# 准备用于计算语言模型损失的标签
# 在语言模型训练中，标签是输入向右移动一位（预测下一个token）
# -100表示在计算损失时忽略该位置（这里忽略提示部分）
labels = torch.LongTensor(
    [
        [-100] * len(prompt_ids) + good_response_ids,
        *[[-100] * len(prompt_ids) + bad_response_ids for bad_response_ids in bad_response_ids_list]
    ]
)# 向右移动一位，因为我们预测的是下一个token
labels
```

```python
labels = labels[:, 1:]  
labels
```

```python
# 创建掩码，用于标识哪些位置参与损失计算（即回答部分）
loss_mask = (labels != -100)
print(loss_mask.shape)
loss_mask
```

```python
# 将-100替换为0，因为在gather操作中-100是无效索引
labels[labels == -100] = 0
print(labels.shape)
labels
```

```python
output = policy_model(input_ids)
for key, value in output.items():  # 如果是ModelOutput对象，可以用output.__dict__.items()
    if hasattr(value, "shape"):
        print(f"{key}: {value.shape}")
    else:
        print(f"{key}: {type(value)}")
```

```python
# ===============================================================================
# 计算策略模型（policy model）的对数概率
# ===============================================================================
# 前向传播，获取每个token位置的预测logits
logits = policy_model(input_ids)["logits"][:, :-1, :]  # 去掉最后一个位置，与label对齐
print(logits.shape)
logits
```

```python
# 将logits转换为对数概率，并提取每个位置上正确token的对数概率
per_token_logps = torch.gather(logits.log_softmax(-1), dim=2, index=labels.unsqueeze(2)).squeeze(2)
per_token_logps
```

```python
# 仅对回答部分（loss_mask=True的位置）求和，得到每个序列的总对数概率
all_logps = (per_token_logps * loss_mask).sum(-1)
all_logps
```

```python
# 分离优质回答和低质量回答的对数概率
policy_good_logps, policy_bad_logps = all_logps[:1], all_logps[1:]
```

```python
policy_good_logps
```

```python
policy_bad_logps
```

```python
# ===============================================================================
# 计算参考模型（reference model）的对数概率
# ===============================================================================
with torch.no_grad():  # 不计算梯度，因为参考模型不需要更新
    # 重复与策略模型相同的步骤
    logits = reference_model(input_ids)["logits"][:, :-1, :]
    print(logits.shape)
    print("logits:\n",logits)
    per_token_logps = torch.gather(logits.log_softmax(-1), dim=2, index=labels.unsqueeze(2)).squeeze(2)
    print(per_token_logps.shape)
    print("per_token_logps:\n",per_token_logps)
    all_logps = (per_token_logps * loss_mask).sum(-1)
    print(all_logps.shape)
    print("all_logps\n",all_logps)
    reference_good_logps, reference_bad_logps = all_logps[:1], all_logps[1:]
    print(reference_good_logps.shape)
    print("reference_good_logps:\n",reference_good_logps)
    print(reference_bad_logps.shape)
    print("reference_bad_logps\n",reference_bad_logps)
```

```python
# ===============================================================================
# 计算DPO损失
# DPO的核心思想：增大策略模型对优质回答的概率，同时减小对低质量回答的概率
# ===============================================================================
# 计算DPO的logits：(策略模型相对于参考模型对好回答的提升) - (对坏回答的提升)
logits = (policy_good_logps - reference_good_logps) - (policy_bad_logps - reference_bad_logps)
# 应用logsigmoid函数并乘以beta控制优化强度，取负值（因为要最小化损失）
loss = -F.logsigmoid(beta * logits).mean()  # 对所有样本取平均

# 输出损失值
print(loss)
```

## ✅ 测试验证

```python
# 验证 DPO 损失性质
import torch
import torch.nn.functional as F

# DPO loss = -log σ(β * (log π(w)/π_ref(w) - log π(l)/π_ref(l)))
# 性质1: 当策略 = 参考时，loss = -log σ(0) = log(2)
beta = 0.1
log_ratio_diff = 0.0  # 策略=参考 → diff=0
loss_init = -F.logsigmoid(torch.tensor(beta * log_ratio_diff)).item()
expected_init = -torch.log(torch.tensor(0.5)).item()  # log(2)
assert abs(loss_init - expected_init) < 1e-6, f"init loss should be log(2)={expected_init}, got {loss_init}"

# 性质2: 胜者概率升高 → loss 减小
log_ratio_diff_good = 5.0  # 胜者 log ratio 远大于败者
loss_good = -F.logsigmoid(torch.tensor(beta * log_ratio_diff_good)).item()
assert loss_good < loss_init, "better separation should reduce loss"

# 性质3: loss 始终非负
assert loss_init > 0 and loss_good > 0, "DPO loss should be non-negative"

# 性质4: β 越大，对相同偏差 loss 变化越大（更敏感）
log_ratio_diff = 2.0
loss_beta_small = -F.logsigmoid(torch.tensor(0.01 * log_ratio_diff)).item()
loss_beta_large = -F.logsigmoid(torch.tensor(1.0 * log_ratio_diff)).item()
assert loss_beta_large < loss_beta_small, "larger beta should be more sensitive"

print(f"✅ DPO 测试通过: 初始loss=log(2)={expected_init:.4f}, 胜者升高loss下降, β调节有效")
```
